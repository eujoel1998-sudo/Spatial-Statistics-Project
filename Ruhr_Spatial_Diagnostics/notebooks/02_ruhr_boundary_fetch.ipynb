```python
import geopandas as gpd
import requests
from io import BytesIO

# FeatureServer layer 0 query endpoint (corrected)
url = "https://services3.arcgis.com/e16K3n7oQnlvAQhE/ArcGIS/rest/services/Regionalverband_Ruhr__RVR/FeatureServer/0/query"

params = {
    "where": "1=1",              # select all features
    "outFields": "*",            # include all attributes
    "f": "geojson",              # request GeoJSON output
    "returnGeometry": "true"     # include geometry
}

# Request GeoJSON
response = requests.get(url, params=params)
response.raise_for_status()  # will raise error if the request fails

# Load GeoJSON into GeoPandas
gdf = gpd.read_file(BytesIO(response.content))

print(gdf.head())
gdf.to_file("RVR_layer.geojson", driver="GeoJSON")  # save locally
```
